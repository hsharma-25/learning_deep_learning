# Operations on tensors

```python
import torch
import numpy as np
```

```python
tensor = torch.rand(3, 5)
```

```python
#slicing, indexing
print(tensor[1])
print(tensor[:, 0])
print(tensor[:, -1])
```

```python
#concatenation
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
#multiplication of tensors
#y1, y2, y3 have same values 

##matrix multiplication  
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)
y3 = torch.rand_like(y1)
torch.matmul(tensor, tensor.T, out=y3)
print(y1)
print(y2)
print(y3)
```

```python
##element wise multiplication
z1 = tensor * tensor
z2 = tensor.mul(tensor)
z3 = torch.rand_like(z1)
torch.mul(tensor, tensor, out=z3)
print(z1)
print(z2)
print(z3)
```

```python
agg = tensor.sum()
agg_item = agg.item()
print(agg_item, type(agg_item))
```

```python
tensor1 = torch.zeros(5, 5)
tensor1.add_(5)
print(tensor1)
```

```python
##tensors on CPU and NumPy arrays can share the same memory space
##changing one will change the other
t = torch.rand(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

```python
n = np.ones(5)
t = torch.from_numpy(n)
np.add(n, 1, out=n)
print(f"n: {n}")
print(f"t: {t}")
```
