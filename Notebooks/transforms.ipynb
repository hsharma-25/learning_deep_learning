# Transforms

- Make data suitable for training
- Use "transform" to modify the features and "target_transform" to modify the labels
- "transform" and "target_transform" are parameteres that every torchvision dataset has

- For FashionMNIST convert features(PIL image) to normalized tensors using "v2.ToImage()" and the labels to one-hot encoded tensors

```python
import torch
import torch.nn.functional as F
from torchvision import datasets
from torchvision.transforms import v2
```

```python
ds = datasets.FashionMNIST(
    root = "data",
    train = True,
    download = True,
    transform = v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale = True)]),    # "v2.ToImage" to convert PIL image into TV tensor and "v2.ToDtype" to convert to float and scaling
    target_transform = v2.Lambda(
        lambda y: F.one_hot(torch.tensor(y), num_classes = 10).float()    # lambda function to convert integer label to one-hot encoded tensor
    )
)
```
