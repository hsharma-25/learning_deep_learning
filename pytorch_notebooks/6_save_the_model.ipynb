# Save and Load Model

```python
import torch
import torchvision.models as models
```

### Workflow:
- vgg16 is a 16 layer deep CNN
- Models store the learned parameters in an internal state dictionary
- Save using the 'torch.save()' method

```python
model = models.vgg16(weights = "IMAGENET1K_V1")
torch.save(model.state_dict(), 'model_weights.pth')
```

### Loading the model
- Create an instance of the model
- Use 'load_state_dict' to load parameters
- 'weights_only = True', limit function execution(is a best practice )

```python
model = models.vgg16()
model.load_state_dict(torch.load('model_weights.pth', weights_only = True))
model.eval()
```
