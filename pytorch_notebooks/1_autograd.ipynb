# Autograd

- Automatic computation of gradients using torch.autograd
- Autograd keeps the record of data(tensors) and all executed operations in a DAG graph
- The leaves are the input tensors and roots are output tensors. The gradient can be computed by tracing from roots to leaves using chain rule
- After each ".backward" call autograd starts populating a new graph

```python
import torch
from torch import nn
```

```python
## simple one-layer neural network
x = torch.ones(5)
y = torch.zeros(3)
w = torch.randn(5, 3, requires_grad = True)
b = torch.randn(3, requires_grad = True)
z = torch.matmul(x, w) + b
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y)
```

```python
print(f"Gradient function for z = {z.grad_fn}")         # grad_fn is the reference to backward propagation function that knows 
print(f"Gradient function for loss = {loss.grad_fn}")   # how to compute function in forward direction and also how to compute its derivative
```

```python
loss.backward()     # the backward pass kicks off when .backward is called
print(w.grad)
print(b.grad)
```

### Disabling Gradient
- Disable gradient tracking if only doing forward pass 
- Speed up computations
- Mark some parameters in the network as frozen

```python
print(z.requires_grad)
with torch.no_grad():
    z = torch.matmul(x, w) + b

print(z.requires_grad)
```
